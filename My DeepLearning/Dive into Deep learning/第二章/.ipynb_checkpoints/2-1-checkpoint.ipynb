## 1. 张量的创建与形状

- `torch.arange(n)`：生成 0 ~ n-1 的一维张量（类似 `range`）。
- `.shape`：查看张量的维度（几行几列），等价于 `size()`。
- `.numel()`：返回张量中元素的总个数。
- `.reshape(r, c)`：改变形状而不改变数据；`-1` 表示该维度自动推断。


```python
x = torch.arange(12)
x
```

```python
x.shape
```

```python
x.numel()
```

```python
X = x.reshape(3, 4)
X
```

## 2. 常见初始化方法

- `torch.zeros(shape)`：全 0 张量。
- `torch.ones(shape)`：全 1 张量。
- `torch.randn(*shape)`：从**标准正态分布**随机抽样（均值 0、方差 1）。


```python
torch.zeros((2, 3, 4))
```

```python
torch.ones((2, 3, 4))
```

```python
torch.randn(3, 4)
```

## 3. 按元素运算（复习 NumPy）

- 张量/数组支持 `+ - * / **` 等按元素（element-wise）运算，`**` 是求幂。
- `np.exp(x)`：对每个元素做指数运算 e^x。
- 这里先用 NumPy 复习数组运算，再过渡到 PyTorch 张量。


```python
import numpy as np
x = np.array([1, 2, 4, 8])
y = np.array([2, 2, 2, 2])
x + y, x - y, x * y, x / y, x ** y  # **运算符是求幂运算
```

```python
np.exp(x)
```

## 4. 拼接、比较与求和

- `torch.cat((A, B), dim=0)`：沿**行方向**（第 0 维）拼接；`dim=1` 沿**列方向**拼接。
- `X == Y`：逐元素比较，返回由 True/False 组成的张量。
- `X.sum()`：把所有元素求和，得到一个标量张量。


```python
X = torch.arange(12, dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
torch.cat((X, Y), dim=0), torch.cat((X, Y), dim=1)
```

```python
X == Y
```

```python
X.sum()
```

## 5. 广播机制（Broadcasting）

- 当两个张量形状不同但**兼容**时，PyTorch 会自动把小的"拉伸"成大的再运算。
- 规则：从最后一维往前对齐，维度为 1 或缺失的会被复制补齐。
- 例：`a` 形状 (2,3,2) 与 `b` 形状 (1,3,2) 相加，`b` 会沿第 0 维复制成 (2,3,2)。


```python
a = torch.arange(12).reshape((2, 3, 2))
b = torch.arange(6).reshape((1, 3, 2))
a, b
```

```python
a + b
```

## 6. 索引与切片

- `X[-1]`：取最后一行；`X[1:3]`：取第 1、2 行（左闭右开，不含 3）。
- `X[i, j] = v`：给单个元素赋值。
- `X[0:2, :] = v`：给一个切片区域整体赋值（整行 / 整列）。


```python
X[-1], X[1:3]
```

```python
X[1, 2] = 9
X
```

```python
X[0:2, :] = 12
X
```

## 7. 内存节省（原地操作）

- `Y = Y + X` 会**新建**一个张量，`id(Y)` 改变（占用新内存）。
- 想复用原内存，用**原地操作**：`X += Y`（或 `Z[:] = X + Y`）。
- 深度学习中张量很大，原地操作能省内存、提升效率。


```python
before = id(Y)
Y = Y + X
id(Y) == before
```

```python
Z = torch.zeros_like(Y)
print('id(Z):', id(Z))
Z[:] = X + Y
print('id(Z):', id(Z))
```

```python
before = id(X)
X += Y
id(X) == before
```

## 8. 与 NumPy 互转 & 提取标量

- `X.numpy()`：把张量转成 NumPy 数组（共享内存）。
- `torch.tensor(A)`：把 NumPy 数组转回张量。
- `a.item()` / `float(a)` / `int(a)`：把**单元素张量**转成 Python 数值（标量）。


```python
A = X.numpy()
B = torch.tensor(A)
type(A), type(B)
```

```python
a = torch.tensor([3.5])
a, a.item(), float(a), int(a)
```
