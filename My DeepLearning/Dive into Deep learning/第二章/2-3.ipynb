## 3. 矩阵（二维张量）与转置

`reshape(5,4)` 造矩阵；`.T` 是转置（行变列）。

```python
A = torch.arange(20).reshape(5, 4)
A
```

```python
A.T
```

## 4. 对称矩阵

`B == B.T` 逐元素比较；对称阵转置后与自己相等。

```python
B = torch.tensor([[1, 2, 3], [2, 0, 4], [3, 4, 5]])
B
```

```python
B == B.T
```

## 5. 多维张量（三维）

```python
X = torch.arange(24).reshape(2, 3, 4)
X
```

## 6. 克隆与哈达玛积（按元素乘）

`A.clone()` 分配新内存避免共享；`A * B` 是逐元素相乘（哈达玛积），**不是**矩阵乘法。

```python
A = torch.arange(20, dtype=torch.float32).reshape(5, 4)
B = A.clone()  # 通过分配新内存，将A的一个副本分配给B
A, A + B
```

```python
A * B
```

```python
a = 2
X = torch.arange(24).reshape(2, 3, 4)
a + X, (a * X).shape
```

```python
x = torch.arange(4, dtype=torch.float32)
x, x.sum()
```

## 9. 点积

`torch.dot(x,y)` 算向量内积，等价于 `sum(x*y)`。

```python
y = torch.ones(4, dtype = torch.float32)
x, y, torch.dot(x, y)
```

```python
torch.sum(x * y)
```

## 10. 矩阵-向量积与矩阵-矩阵积

`torch.mv(A,x)` 矩阵乘向量，`torch.mm(A,B)` 矩阵乘矩阵——这正是线性层的核心运算。

```python
A.shape, x.shape, torch.mv(A, x)
```

```python
B = torch.ones(4, 3)
torch.mm(A, B)
```
